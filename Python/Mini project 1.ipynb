```python
lines = []

try:
    with open("users.txt", "r", encoding="utf-8") as file:
        for line in file:
            line = line.strip()
            if line:               # пропускаем пустые строки
                lines.append(line)

except FileNotFoundError:
    print("the file that you would like to read is not found")

except Exception:
    print("some error occured")
```

```python
# Задание 2
```

```python
class Users:
    def __init__(self, name, surname, birth_year, email):
        self.name = name
        self.surname = surname
        self.birth_year = int(birth_year)
        self.email = email

    def getAge(self):
        return 2023 - self.birth_year

    def __str__(self):
        return f"{self.name} {self.surname}, {self.birth_year}, {self.email}, {self.getAge()}"
```

```python
# Задание 3
```

```python
users_list = []

for s in lines:
    parts = s.split(",")

    name = parts[0].strip()
    surname = parts[1].strip()
    email = parts[2].strip()
    birth_year = parts[3].strip()

    user = Users(name, surname, birth_year, email)
    users_list.append(user)
```

```python
for user in users_list:
    print(user)
```

```python
# Задание 4
```

```python
def filter(users, low=30, high=50):
    result = []

    for user in users:
        age = user.getAge()
        if age > low and age < high:
            result.append(user)
            print(f"{user.email},{age}")

    return result
```

```python
_ = filter(users_list)
```
